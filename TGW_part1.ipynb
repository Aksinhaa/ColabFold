# Ancient DNA Workshop: Trimming and Mapping

In this workshop, we will:
- Perform quality control using FastQC
- Trim adapters using AdapterRemoval
- Map reads to a reference genome using BWA
- Process alignment files using Samtools


```python
%%bash
# Install Miniconda
wget -q https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh
bash Miniconda3-latest-Linux-x86_64.sh -b -p /usr/local/miniconda

# Initialize conda
source /usr/local/miniconda/etc/profile.d/conda.sh

# Accept Terms of Service
conda tos accept --override-channels --channel https://repo.anaconda.com/pkgs/main
conda tos accept --override-channels --channel https://repo.anaconda.com/pkgs/r

# Add channels
conda config --add channels defaults
conda config --add channels bioconda
conda config --add channels conda-forge
```

```python
%%bash
source /usr/local/miniconda/etc/profile.d/conda.sh

conda create -y -n ngs_env
```

```python
%%bash
source /usr/local/miniconda/etc/profile.d/conda.sh
conda activate ngs_env

conda install -y bwa samtools fastqc adapterremoval
```

```python
%%bash

# Placeholder file names
READ1="read1.fq"
READ2="read2.fq"
REF="reference.fasta"

echo "Inputs:"
echo $READ1 $READ2 $REF
```

```python
%%bash
source /usr/local/miniconda/etc/profile.d/conda.sh
conda activate ngs_env

fastqc read1.fq read2.fq
```

```python
%%bash
source /usr/local/miniconda/etc/profile.d/conda.sh
conda activate ngs_env

bwa index reference.fasta
```

```python
%%bash
source /usr/local/miniconda/etc/profile.d/conda.sh
conda activate ngs_env

AdapterRemoval \
--file1 read1.fq \
--file2 read2.fq \
--basename trimmed \
--trimns \
--trimqualities \
--minlength 25 \
--collapse
```
